```python
import os
import pandas as pd
import matplotlib.pyplot as plt


results_path = '/Users/macbook/code/Soumiabenhamou90/Brain-Tumor-Detection/brainTumorDetection/notebooks/runs/detect/train3/results.csv'

results = pd.read_csv(results_path)

plt.figure()
plt.plot(results['epoch'], results['train/box_loss'], label='train box_loss')
plt.plot(results['epoch'], results['val/box_loss'], label='val lbox_loss', c='red')
plt.grid()
plt.title('Loss vs epochs')
plt.ylabel('box-loss')
plt.xlabel('epochs')
plt.legend()


plt.figure()
plt.plot(results['epoch'], results['metrics/recall(B)'] * 100)
plt.grid()
plt.title('Validation recall vs epochs')
plt.ylabel('Recall (%)')
plt.xlabel('epochs')

plt.show()
```
